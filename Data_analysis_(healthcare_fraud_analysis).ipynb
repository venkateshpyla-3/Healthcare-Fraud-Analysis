```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "../data/cleaned_health_claims.csv",
    parse_dates=["Claim_Date", "Service_Date", "Policy_Expiration_Date"]
)

df.shape
```

```python
claim_bins = pd.cut(
    df["Claim_Amount"],
    bins=[0, 500000, 1000000, 1500000, 2000000, float("inf")],
    labels=["Below 500K", "500K-1M", "1M-1.5M", "1.5M-2M", "2M+"]
)

claim_amount_fraud = (
    df.groupby(claim_bins, observed=True)["Is_Fraudulent"]
    .mean()
    .mul(100)
)

claim_amount_fraud
```

```python
claim_amount_fraud.plot(kind="bar")

plt.xlabel("Claim Amount Range")
plt.ylabel("Fraud Rate (%)")
plt.title("Fraud Rate by Claim Amount Range")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
age_bins = pd.cut(
    df["Patient_Age"],
    bins=[-1, 17, 30, 45, 60, float("inf")],
    labels=["0-17", "18-30", "31-45", "46-60", "61+"]
)

age_fraud = (
    df.groupby(age_bins, observed=True)["Is_Fraudulent"]
    .mean()
    .mul(100)
)

age_fraud
```

```python
service_fraud = (
    df.groupby("Service_Type")["Is_Fraudulent"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

service_fraud
```

```python
service_fraud.plot(kind="bar")

plt.xlabel("Service Type")
plt.ylabel("Fraud Rate (%)")
plt.title("Fraud Rate by Service Type")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
df["Claim_Month"] = df["Claim_Date"].dt.to_period("M")

monthly_fraud = (
    df.groupby("Claim_Month")["Is_Fraudulent"]
    .mean()
    .mul(100)
)

monthly_fraud.head()
```

```python
monthly_fraud.plot(kind="line", figsize=(10, 5))

plt.xlabel("Month")
plt.ylabel("Fraud Rate (%)")
plt.title("Monthly Fraud Rate Trend")
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
numeric_cols = [
    "Claim_Amount",
    "Patient_Age",
    "Length_of_Stay_Days",
    "Deductible_Amount",
    "CoPay_Amount",
    "Provider_Patient_Distance_Miles",
    "Is_Fraudulent"
]

correlation = df[numeric_cols].corr()

correlation["Is_Fraudulent"].sort_values(ascending=False)
```

```python
import seaborn as sns

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
provider_fraud = (
    df.groupby("Provider_Type")["Is_Fraudulent"]
    .agg(["count", "sum", "mean"])
)

provider_fraud["fraud_rate"] = provider_fraud["mean"] * 100

provider_fraud.sort_values(
    "fraud_rate",
    ascending=False
)
```

```python
provider_fraud["fraud_rate"].sort_values(
    ascending=False
).plot(kind="bar")

plt.xlabel("Provider Type")
plt.ylabel("Fraud Rate (%)")
plt.title("Fraud Rate by Provider Type")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
combo_fraud = (
    df.groupby(["Provider_Type", "Provider_Specialty"])
      ["Is_Fraudulent"]
      .agg(["count", "sum", "mean"])
)

combo_fraud["fraud_rate"] = combo_fraud["mean"] * 100

combo_fraud = combo_fraud[
    combo_fraud["count"] >= 50
].sort_values("fraud_rate", ascending=False)

combo_fraud.head(10)
```

```python
combo_plot = combo_fraud.head(10).reset_index()

combo_plot["combination"] = (
    combo_plot["Provider_Type"]
    + " - "
    + combo_plot["Provider_Specialty"]
)

combo_plot.plot(
    x="combination",
    y="fraud_rate",
    kind="bar",
    legend=False
)

plt.xlabel("Provider Type - Specialty")
plt.ylabel("Fraud Rate (%)")
plt.title("Top Provider Type-Specialty Combinations by Fraud Rate")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
diagnosis_fraud = (
    df.groupby("Diagnosis_Code")["Is_Fraudulent"]
      .agg(["count", "sum", "mean"])
)

diagnosis_fraud["fraud_rate"] = diagnosis_fraud["mean"] * 100

diagnosis_fraud = (
    diagnosis_fraud[diagnosis_fraud["count"] >= 50]
    .sort_values("fraud_rate", ascending=False)
)

diagnosis_fraud.head(10)
```

```python
procedure_fraud = (
    df.groupby("Procedure_Code")["Is_Fraudulent"]
      .agg(["count", "sum", "mean"])
)

procedure_fraud["fraud_rate"] = procedure_fraud["mean"] * 100

procedure_fraud = (
    procedure_fraud[procedure_fraud["count"] >= 50]
    .sort_values("fraud_rate", ascending=False)
)

procedure_fraud.head(10)
```

## Key EDA Findings

- Overall fraud rate: 24.64%.
- Fraudulent claims have a higher average claim amount than non-fraudulent claims.
- Cardiology and Orthopedics have the highest specialty-level fraud rates (~46%).
- Specialist Office has a 49.68% fraud rate.
- Specialist Office + Cardiology and Specialist Office + Orthopedics exceed 80% fraud.
- Provider-patient distances of 500+ miles show higher fraud rates (~31–32%).
- Claim amounts between $500K and $1.5M show elevated fraud rates.
- Elective and Urgent admissions have fraud rates around 32%.
- Procedure codes 99203 and 99213 show fraud rates around 40%.
- Diagnosis codes R05 and J02.9 show fraud rates around 38%.
- Patient age shows little variation in fraud rate.
- Monthly fraud rates fluctuate without a clear long-term trend.
